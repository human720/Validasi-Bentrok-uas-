```python
# =========================
# VALIDASI BENTROK JADWAL
# =========================
def cek_bentrok(id_fasilitas, tanggal, mulai, selesai):
    """Mengembalikan True jika ada booking yang waktunya bertabrakan."""
    with koneksi() as conn:
        cur = conn.cursor()
        cur.execute("""
            SELECT COUNT(*) FROM booking
            WHERE id_fasilitas = ?
              AND tanggal = ?
              AND status_booking != 'DIBATALKAN'
              AND jam_mulai < ?
              AND jam_selesai > ?
        """, (id_fasilitas, tanggal, selesai, mulai))
        return cur.fetchone()[0] > 0
```
